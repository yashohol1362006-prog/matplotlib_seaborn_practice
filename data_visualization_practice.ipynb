(or box and whisker plot) shwos the distribution of quantitative data in a way that facilitates comperison between variables or across levels of categorial variable. The box shows the quartiles of the dataset while the whiskers extend to show the rest of the distribution, except for the poiints that are determined to be 'outliers' using a method that funtion of the inter-quartile range.

```python
import numpy as np 
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
x = np.outer(np.linspace(-2, 2, 10), np.ones(10))
y = x.copy().T
z = np.cos(x**2 + y**3)

fig = plt.figure()
ax = plt.axes(projection='3d')
ax.plot_surface(x, y, z, cmap='viridis', edgecolor='green')
ax.set_title('Surface Plot')
plt.show()
```
